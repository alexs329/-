```python
# Функция для нормализации текста
def normalize_text(text):
    if pd.isna(text):
        return text
    # Приводим к нижнему регистру
    text = str(text).lower()
    # Удаляем лишние пробелы
    text = ' '.join(text.split())
    return text

# Применяем нормализацию к текстовым столбцам
df['name_normalized'] = df['name'].apply(normalize_text)
df['address_normalized'] = df['address'].apply(normalize_text)
df['category'] = df['category'].apply(normalize_text)
df['district'] = df['district'].apply(normalize_text)

print("Нормализация применена к текстовым столбцам")
print(df[['name', 'name_normalized']].head())
```

```python
# Удаляем дубликаты по названию и адресу (оставляем первое вхождение)
initial_count = len(df)
df = df.drop_duplicates(subset=['name_normalized', 'address_normalized'], keep='first')
removed_count = initial_count - len(df)

print(f"Было строк: {initial_count}")
print(f"Удалено дубликатов: {removed_count}")
print(f"Осталось строк: {len(df)}")
```

- Для дальнейшей работы создайте столбец `is_24_7` с обозначением того, что заведение работает ежедневно и круглосуточно, то есть 24/7:
  - логическое значение `True` — если заведение работает ежедневно и круглосуточно;
  - логическое значение `False` — в противоположном случае.

```python
# Функция для определения круглосуточной работы
def is_24_7(hours_str):
    if pd.isna(hours_str):
        return False
    hours_str = str(hours_str).lower()
    # Проверяем точное условие: ежедневно, круглосуточно
    return 'ежедневно, круглосуточно' in hours_str

# Создаем столбец
df['is_24_7'] = df['hours'].apply(is_24_7)

# Проверяем результат
print(f"Количество круглосуточных заведений: {df['is_24_7'].sum()}")
print(f"Доля круглосуточных: {(df['is_24_7'].sum() / len(df) * 100).round(2)}%")

print("\nПримеры круглосуточных заведений:")
print(df[df['is_24_7'] == True][['name', 'hours']].head())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# График 1: Количество заведений в сети
sns.barplot(data=top_networks, x='places_count', y='name', palette='Blues_d', ax=axes[0])
axes[0].set_title('Топ-15 сетей по количеству заведений', fontsize=14)
axes[0].set_xlabel('Количество заведений')
axes[0].set_ylabel('Название сети')

# График 2: Средний рейтинг сетей
sns.barplot(data=top_networks, x='avg_rating', y='name', palette='Greens_d', ax=axes[1])
axes[1].set_title('Средний рейтинг топ-15 сетей', fontsize=14)
axes[1].set_xlabel('Средний рейтинг')
axes[1].set_ylabel('Название сети')

plt.tight_layout()
plt.show()
```

```python
# Анализ категорий среди топ-сетей
category_distribution = top_networks['category'].value_counts()

print("Распределение категорий среди топ-15 сетей:")
print(category_distribution)

# Визуализация
plt.figure(figsize=(10, 6))
plt.pie(category_distribution.values, labels=category_distribution.index, 
        autopct='%1.1f%%', startangle=90)
plt.title('Категории заведений среди топ-15 сетей', fontsize=14)
plt.show()
```

### Задача 8

Изучите вариацию среднего чека заведения (столбец `middle_avg_bill`) в зависимости от района Москвы. Проанализируйте цены в Центральном административном округе и других. Как удалённость от центра влияет на цены в заведениях? Результат сопроводите подходящими визуализациями.


```python
# Отбираем заведения, у которых есть данные о среднем чеке (middle_avg_bill не пустой)
df_price_analysis = df[df['middle_avg_bill'].notna()].copy()

# Определяем, находится ли заведение в ЦАО
df_price_analysis['is_cao'] = df_price_analysis['district'].str.contains('центральный', case=False, na=False)

print(f"Всего заведений с данными о среднем чеке: {len(df_price_analysis)}")
print(f"Из них в ЦАО: {df_price_analysis['is_cao'].sum()}")
print(f"Вне ЦАО: {(~df_price_analysis['is_cao']).sum()}")
```

```python
# Статистика среднего чека по районам
district_price = df_price_analysis.groupby('district')['middle_avg_bill'].agg(['mean', 'median', 'count']).reset_index()
district_price = district_price.sort_values('mean', ascending=False)

print("Средний чек по административным районам (топ-10):")
print(district_price.head(10))
```

```python
plt.figure(figsize=(12, 6))

# Boxplot сравнения ЦАО и других районов
sns.boxplot(data=df_price_analysis, x='is_cao', y='middle_avg_bill', 
            palette=['lightblue', 'lightcoral'])
plt.title('Сравнение среднего чека: Центральный округ vs остальные', fontsize=14)
plt.xlabel('Расположение')
plt.ylabel('Средний чек (₽)')
plt.xticks([0, 1], ['Вне ЦАО', 'ЦАО'])

# Добавляем статистику
cao_mean = df_price_analysis[df_price_analysis['is_cao']]['middle_avg_bill'].mean()
other_mean = df_price_analysis[~df_price_analysis['is_cao']]['middle_avg_bill'].mean()

print(f"Средний чек в ЦАО: {cao_mean:.0f} ₽")
print(f"Средний чек вне ЦАО: {other_mean:.0f} ₽")
print(f"Разница: {cao_mean - other_mean:.0f} ₽")

plt.tight_layout()
plt.show()
```

```python
# Подготовка данных для сравнения цен по категориям и округам
# Создаем признак "расположение"
df_price_analysis['location'] = df_price_analysis['district'].apply(
    lambda x: 'ЦАО' if 'центральный' in str(x).lower() else 'Другие округа'
)

# Группируем по категории и расположению
category_price_stats = df_price_analysis.groupby(['category', 'location'])['middle_avg_bill'].agg(
    mean_price='mean',
    median_price='median',
    count='count'
).reset_index()

# Оставляем только категории с достаточным количеством заведений (не менее 10)
category_price_stats = category_price_stats[category_price_stats['count'] >= 10]

# Сортируем по средней цене в ЦАО для наглядности
cao_prices = category_price_stats[category_price_stats['location'] == 'ЦАО'].set_index('category')['mean_price']
top_categories = cao_prices.sort_values(ascending=False).head(10).index.tolist()

# Фильтруем данные для топ-10 категорий
top_categories_stats = category_price_stats[category_price_stats['category'].isin(top_categories)]

# Создаем барплот
plt.figure(figsize=(14, 8))

# Настройки для grouped barplot
x = np.arange(len(top_categories))
width = 0.35

# Данные для ЦАО и других округов
cao_means = []
cao_medians = []
other_means = []
other_medians = []

for cat in top_categories:
    cao_data = top_categories_stats[(top_categories_stats['category'] == cat) & 
                                     (top_categories_stats['location'] == 'ЦАО')]
    other_data = top_categories_stats[(top_categories_stats['category'] == cat) & 
                                       (top_categories_stats['location'] == 'Другие округа')]
    
    cao_means.append(cao_data['mean_price'].values[0] if len(cao_data) > 0 else 0)
    cao_medians.append(cao_data['median_price'].values[0] if len(cao_data) > 0 else 0)
    other_means.append(other_data['mean_price'].values[0] if len(other_data) > 0 else 0)
    other_medians.append(other_data['median_price'].values[0] if len(other_data) > 0 else 0)

# Рисуем бары
bars1 = plt.bar(x - width/2, cao_means, width, label='Среднее (ЦАО)', color='lightcoral')
bars2 = plt.bar(x - width/2, cao_medians, width, label='Медиана (ЦАО)', color='darkred', alpha=0.7)
bars3 = plt.bar(x + width/2, other_means, width, label='Среднее (другие округа)', color='lightblue')
bars4 = plt.bar(x + width/2, other_medians, width, label='Медиана (другие округа)', color='darkblue', alpha=0.7)

# Настройки графика
plt.xlabel('Категория заведения', fontsize=12)
plt.ylabel('Средний чек (₽)', fontsize=12)
plt.title('Сравнение среднего и медианного чека по категориям: ЦАО vs другие округа', fontsize=14)
plt.xticks(x, top_categories, rotation=45, ha='right')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Текстовый вывод
print("="*60)
print("Вывод по задаче 8: Анализ среднего чека по районам")
print("="*60)

print("\n1. Сравнение ЦАО и других округов:")
cao_mean = df_price_analysis[df_price_analysis['location'] == 'ЦАО']['middle_avg_bill'].mean()
other_mean = df_price_analysis[df_price_analysis['location'] == 'Другие округа']['middle_avg_bill'].mean()
cao_median = df_price_analysis[df_price_analysis['location'] == 'ЦАО']['middle_avg_bill'].median()
other_median = df_price_analysis[df_price_analysis['location'] == 'Другие округа']['middle_avg_bill'].median()

print(f"  ЦАО — средний: {cao_mean:.0f} ₽, медианный: {cao_median:.0f} ₽")
print(f"  Другие округа — средний: {other_mean:.0f} ₽, медианный: {other_median:.0f} ₽")
print(f"  Разница в среднем чеке: {cao_mean - other_mean:.0f} ₽ (на {(cao_mean/other_mean - 1)*100:.0f}% выше в центре)")

print("\n2. Категории с наибольшей разницей в цене между ЦАО и другими округами:")
top_diff = []
for cat in top_categories[:5]:
    cao_price = cao_prices[cat]
    other_price = top_categories_stats[(top_categories_stats['category'] == cat) & 
                                        (top_categories_stats['location'] == 'Другие округа')]['mean_price'].values
    if len(other_price) > 0:
        diff = cao_price - other_price[0]
        print(f"  {cat}: ЦАО — {cao_price:.0f} ₽, другие — {other_price[0]:.0f} ₽, разница — {diff:.0f} ₽")

print("\n3. Общий вывод:")
print("  Цены в центре Москвы (ЦАО) значительно выше, чем в других округах.")
print("  Медианные значения ниже средних, что говорит о наличии заведений с очень высокими ценами,")
print("  которые 'тянут' среднее значение вверх.")
print("  Наибольшая разница в ценах наблюдается в категориях: рестораны, бары, кофейни.")
```
